### 1.Gráfico de Uma Função Linear

Gráfico para a função y = 2x + 3 para valores de x no intervalo de 5 a 10

```python
# Gera os dados
x = np.linspace(5, 10, 100)
y = 2 * x + 3

# Plot
plt.plot(x,y)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Gráfico da Função Linear y = 2x + 3')
plt.grid(True)

plt.show()
```
